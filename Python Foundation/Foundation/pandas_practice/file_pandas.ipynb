```python
import pandas as pd
data = {
    "Student_ID": [101, 102, 103, 104, 105,106,107,108],
    "Name": ["Arun", "Priya", "Karthik", "Meena", "Rahul","Abdul","Hanvika","Bharath"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT","CSE","IT","ECE"],
    "Attendance": [82, 91, 76, 68, 88, 57, 87, 79],
    "Python": [85, 92, 76, 65, 58, 87, 90, 79],
    "SQL": [78, 88, 82, 70, 91, 85, 71, 76],
    "Excel": [90, 95, 80, 72, 86, 65, 87, 57]
}
df = pd.DataFrame(data)
print(df)
```

```python
#avg of python , sql , excel
df["Average"]=((df["Python"]+df["SQL"]+df["Excel"])//3)
df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

dep_py_avg = df.groupby("Department")["Python"].mean()

dep_py_avg.plot(kind="bar", color="blue")

plt.title("Department Wise Python Average")
plt.ylabel("Average Python Mark")
plt.xlabel("Department")

plt.show()
```
