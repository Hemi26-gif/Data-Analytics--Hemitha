```python
b1 = np.array([                            #Axis 0 ----> x --> Row 
    [56,78,89],                               #Axis 1 ---> Y --> Column
    [34,45,67]]
    )
b2 = np.array([
    [90,87,69],
    [29,35,50]]
    )
comb1 = np.concatenate((b1,b2),axis=1)
print(comb1)
print("-----------")
comb0 = np.concatenate((b1,b2),axis=0)
comb0
```

```python
mark = np.array([
[89,67,90],
[92,83,85],
[98,56,43]
])
print(np.where(mark==85))
print(mark)
```

```python

np.where(mark>=80) # gives indexes
print(mark[np.where(mark>=80)]) # gives exact values
print(mark)
```

```python
mark = np.array([
[89,67,90],
[92,83,85],
[98,56,43]
])
np.sort(mark)
np.sort(mark,axis=1)
np.sort(mark,axis=0)
print(np.sort(mark)[::-1])
# print(mark)
```
