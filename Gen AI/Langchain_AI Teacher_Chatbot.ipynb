```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=GEMINI_API_KEY,
    temperature=0.7
)

print("Gemini connected through LangChain!")
```

```python
from langchain_core.prompts import PromptTemplate

template = """
You are an experienced AI trainer.

Explain the following topic: {topic}

Follow this format:

1. Definition
2. Simple explanation
3. Three real-world examples
4. Practical applications
5. Five interview questions

Use simple English suitable for beginners.
"""

prompt = PromptTemplate.from_template(template)

print("Prompt template created successfully!")
```

```python
chain = prompt | llm

print("Prompt and Gemini connected successfully!")
```

```python
response = chain.invoke({
    "topic": "Machine Learning"
})

print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

print("Output parser created!")
```

```python
chain = prompt | llm | output_parser

print("Complete LangChain workflow created!")
```

```python
result = chain.invoke({
    "topic": "Generative AI"
})

print(result)
print("\nOutput type:", type(result))
```

```python
print("===== LANGCHAIN AI TEACHER =====")

while True:

    topic = input("\nEnter a topic (or exit): ")

    if topic.lower() == "exit":
        print("Thank you!")
        break

    try:

        result = chain.invoke({
            "topic": topic
        })

        print("\n===== AI GENERATED RESPONSE =====")
        print(result)

    except Exception as e:
        print("Error:", str(e))
```
